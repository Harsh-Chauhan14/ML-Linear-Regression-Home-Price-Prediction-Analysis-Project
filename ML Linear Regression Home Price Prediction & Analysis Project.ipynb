```python
data["Distance"].fillna(0,inplace=True)
data["Bedroom2"].fillna(0,inplace=True)
data["Bathroom"].fillna(0,inplace=True)
data["Car"].fillna(0,inplace=True)
data["Propertycount"].fillna(0,inplace=True)
```

```python
data.fillna(data[["Landsize","BuildingArea"]].mean(),inplace=True)
```

```python
data=pd.get_dummies(data,columns=["Method"])
```

```python
data=pd.get_dummies(data,columns=["Suburb"])
```

```python
data=pd.get_dummies(data,columns=["Type","SellerG","CouncilArea","Regionname"])
```

```python
x=data.drop("Price",axis=1)
y=data["Price"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
yp=model.predict(x_test)
```

```python
from sklearn.metrics import r2_score,root_mean_squared_error,mean_absolute_error,mean_squared_error
```

```python
r2_score(yp,y_test)
```

```python
root_mean_squared_error(yp,y_test)
```

```python
model.score(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
mean_absolute_error(yp,y_test)
```

```python
mean_squared_error(yp,y_test)
```
